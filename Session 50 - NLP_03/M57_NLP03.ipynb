Preprocessing

```python
corpus = []

for i in range(len(sentences)):
  review = re.sub('[^A-z]', ' ', sentences[i])

  review = review.lower()

  review = review.split()

  review = [stemmer.stem(word) for word in review if word not in stopwords.words('english')]

  review = ' '.join(review)
  corpus += [review]
```

```python
corpus
```

**TF**

```python
def term_frequency(word, doc):
  return doc.count(word) / len(doc)
```

```python
for doc in corpus:
  for word in doc.split():
    print(word, ":" ,term_frequency(word, doc))
  print('==========================')
```

**Inverse Document Frequency (IDF)**

```python
import math

def inverse_doc_frequency(word, corpus):
  count_of_documents = len(corpus)
  count_of_documents_with_word = sum([1 for doc in corpus if word in doc])
  IDF = math.log10(count_of_documents / (count_of_documents_with_word + 1) + 1)
  return IDF
```

```python
for doc in corpus:
  for word in doc.split():
    print(word, ':', inverse_document_frequency(word, corpus))
```

**TF-IDF**

```python
tf_idf = {word: (term_frequency(word, doc) * inverse_document_frequency(word, corpus)) for doc in corpus for word in doc.split()}
```
