```python
for i in range(img.shape[0] // 2):
    temp = img[i]
    img[i] = img[img.shape[0] - i - 1]
    img[len(img) - i - 1] = temp

imshow(img)
```

```python
for i in range(img2.shape[1] // 2):
    temp = img2[i]
    img2[i] = img2[img2.shape[1] - i - 1]
    img2[img2.shape[1] - i - 1] = temp

imshow(img2)
```
