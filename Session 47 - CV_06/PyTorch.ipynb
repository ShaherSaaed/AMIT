```python
from torchvision import datasets, transforms # MNIST
from torch.utils.data import DataLoader
```

```python
data = datasets.MNIST(root='./data', train=True, download=True, transform=transforms.ToTensor())
```

```python
loader = torch.utils.data.DataLoader(data, batch_size=60000, shuffle=True)
```

```python
images, _ = next(iter(loader))
```

```python
images
```

```python
mean = images.mean()
std = images.std()
```

```python
print(mean)
print(std)
```

```python
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((mean,), (std))
])
```

```python
train_data = datasets.MNIST(root='./data', train=True, download=True, transform=transform)
test_data = datasets.MNIST(root='./data', train=False, download=True, transform=transform)
```

```python
train_loader = DataLoader(train_data, batch_size=32, shuffle=True)
test_loader = DataLoader(test_data, batch_size=32, shuffle=False)
```

NN

```python
import torch.nn as nn
import torch.nn.functional as f
```

```python
class Net(nn.Module):
  def __init__(self):
    super(Net, self).__init__()
    self.fc1 = nn.Linear(28 * 28, 128)
    self.fc2 = nn.Linear(128, 64)
    self.fc3 = nn.Linear(64, 10)

  def forward(self, x):
    x = x.view(-1, 28*28)
    x = f.relu(self.fc1(x))
    x = f.relu(self.fc2(x))
    x = self.fc3(x)
    return x


model = Net()
```

```python
import torch.optim as optim
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)
```

```python
epochs = 10

for epoch in range(epochs):
    running_loss = 0.0
    for images, labels in train_loader:
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()

    print(f"Epoch {epoch+1} - Loss: {running_loss / len(train_loader):.4f}")
```

```python
correct = 0
total = 0

with torch.no_grad():
  for images, labels in test_loader:
    outputs = model(images)
    _, predicted = torch.max(outputs.data, 1) # >> [0.1, 0.05, 0.3, "0.6", 0.0, 0.2, 0.1, 0.1, 0.3, 0.15]
    total += labels.size(0)
    correct += (predicted == labels).sum().item()

accuracy = 100 * correct / total
print(f'Accuracy: {accuracy:.2f}')
```
