Stop Words Removal in Arabic and English using NLTK

```python
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
```

Removing Stopwords from Arabic Text

```python
stop_words = set(stopwords.words('arabic'))
```

```python
stop_words
```

```python
arabic_text = 'قامت الشركة باصدار بيانات تفصيلية حول أدائها في الربع الأول من العام الحالي'
```

```python
from nltk.tokenize import word_tokenize
```

```python
words = word_tokenize(arabic_text)
```

```python
words
```

```python
filtered_words = [word for word in words if word not in stop_words]
```

```python
filtered_text = ' '.join(filtered_words)
```

```python
arabic_text
```

```python
filtered_text
```

Stemming and Lemmatization in NLTK

```python
from nltk.stem import PorterStemmer
```

```python
stemmer = PorterStemmer()
```

```python
words = ['walking', 'jumps', 'jumped', 'jumping']
```

```python
for word in words:
  stemmed_word = stemmer.stem(word)
  print(word, '>>>', stemmed_word)
```

```python
text = 'The quick brown foxes jumped over the lazy dogs'
```

```python
stemmed_words = [stemmer.stem(word) for word in word_tokenize(text)]
```

```python
stemmed_words
```

Part Of Speech (POS)

```python
text = 'ELon flew to Mars yesterday, He carried biryani masala with him'
```

```python
import spacy
```

```python
!python -m spacy download en_core_web_sm
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp(text)
```

```python
text = 'ELon flew to Mars yesterday, He carried biryani masala with him'
```

```python
print(doc)
```

```python
for token in doc:
  print(token, '>>>', token.pos_, '>>>', spacy.explain(token.pos_))
```

```python
earnings_text = """
Microsoft Corp. today announced the following results for the quarter ended December 31, 2021, as compared to the corresponding period of last fiscal year:
- Revenue was $51.7 billion and increased 20%
- Operating income was $22.2 billion and increased 24%
- Net income was $18.8 billion and increased 21%
- Diluted earnings per share was $2.48 and increased 22%
"""
```

```python
doc = nlp(earnings_text)
```

```python
filtered_tokens = []

for token in doc:
  if token.pos_ not in ['SPACE', 'PUNCT', 'X']: # X >>> [@, #, etc...]
     filtered_tokens.append(token)
```

```python
print(filtered_tokens)
```

```python
count = doc.count_by(spacy.attrs.POS)
```

```python
count
```

```python
for k, v in count.items():
  print(doc.vocab[k].text, '>>>', v)
```

Named Entity Recognition (NER)

```python
nlp = spacy.load('en_core_web_sm')
```

```python
text = 'Tesla Inc is going to acquire Twitter for $45 Billion'
```

```python
doc = nlp(text)
```

```python
doc.ents
```

```python
for ent in doc.ents:
  print(ent.text, '>>>', ent.label_, '>>>', spacy.explain(ent.label_))
```

```python
from spacy import displacy
```

```python
displacy.render(doc, style='ent')
```

```python
nlp.pipe_labels['ner']
```

Regular Expressions ([Regex](https://regex101.com/))

```python
import re
```

```python
text = '''
Elon Musk's phone number is 9991116666, call him if you have any questions on Dogecoin.
Tesla's revenue is 40 billion.
Tesla's CFO number (999)-333-7777.
'''
```

```python
# ['999', '111', '666', '999', '333', '777']
```

```python
pattern = r'\d{3}'
print(re.findall(pattern, text))
```

```python
# ['(999)-333-7777']
```

```python
pattern = r'\(\d{3}\)-\d{3}-\d{4}'
print(re.findall(pattern, text))
```

```python
pattern = r'\(\d{3}\)-\d{3}-\d{4}|\d{10}'
print(re.findall(pattern, text))
```
