Translation

```python
from skimage import transform
```

```python
matrix = np.array([[1, 0, 100],
                   [0, 1, 100],
                   [0, 0, 1]])
```

```python
tform = transform.SimilarityTransform(matrix)
```

```python
res = transform.warp(image=img, inverse_map=tform)
```

```python
imshow(res)
```

```python
tform2 = transform.SimilarityTransform(translation=(100, 100))
```

```python
tform2
```

```python
res = transform.warp(image=img, inverse_map=tform2)
```

```python
imshow(res)
```

Fix  the Scaling

```python
img_center = np.array(img.shape) // 2
```

```python
img_center
```

```python
scale_center = transform.SimilarityTransform(translation=(-img_center[0], -img_center[1]))
scale = transform.SimilarityTransform(scale=0.5)
scale_inverse = transform.SimilarityTransform(translation=(img_center[0], img_center[1]))
```

```python
tform3 = scale_center + scale + scale_inverse
```

```python
result = transform.warp(img, tform3.inverse)
```

```python
imshow(result)
```
